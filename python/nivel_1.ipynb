# 📘 Aula 01 Variaveis E Tipos

## NÍVEL 1 — Fundamentos | Aula 1: Variáveis e Tipos de Dados

Uma variável é uma "caixa etiquetada" na memória do computador.

```python
nome = "Ana"         # str   → texto, sempre entre aspas
idade = 25           # int   → número inteiro
altura = 1.68        # float → número com casas decimais
ativo = True         # bool  → verdadeiro (True) ou falso (False)

# Exibindo valores
print(nome)
print(idade)
print(altura)
print(ativo)

# Descobrindo o tipo de uma variável
print(type(nome))    # <class 'str'>
print(type(idade))   # <class 'int'>
print(type(altura))  # <class 'float'>
print(type(ativo))   # <class 'bool'>

# f-string: montando frases com variáveis (a forma mais usada no dia a dia)
print(f"{nome} tem {idade} anos e {altura}m de altura.")

# Múltiplas atribuições na mesma linha
x, y, z = 1, 2, 3
print(x, y, z)

# Reatribuição: a variável pode mudar de valor
contador = 0
contador = contador + 1   # agora contador vale 1
contador += 1             # atalho para o mesmo acima — agora vale 2
print(contador)

# Conversão entre tipos (casting)
numero_texto = "42"
numero_inteiro = int(numero_texto)   # "42" → 42
numero_decimal = float(numero_texto) # "42" → 42.0
texto_do_numero = str(99)            # 99  → "99"
print(type(numero_inteiro), numero_inteiro)
```

# 📘 Aula 02 Operadores

## NÍVEL 1 — Fundamentos | Aula 2: Operadores

## 1. Operadores Aritméticos

```python
a = 10
b = 3

print(a + b)    # 13  → soma
print(a - b)    # 7   → subtração
print(a * b)    # 30  → multiplicação
print(a / b)    # 3.333... → divisão real (sempre retorna float)
print(a // b)   # 3   → divisão inteira (descarta o decimal)
print(a % b)    # 1   → resto da divisão (módulo)
print(a ** b)   # 1000 → potência (10³)

# Caso real: calcular o valor total de uma venda
preco_unitario = 25.90
quantidade = 4
desconto = 0.10  # 10%

subtotal = preco_unitario * quantidade
valor_desconto = subtotal * desconto
total = subtotal - valor_desconto

print(f"Subtotal: R${subtotal:.2f}")         # :.2f → 2 casas decimais
print(f"Desconto: R${valor_desconto:.2f}")
print(f"Total: R${total:.2f}")
```

## 2. Operadores de Comparação

Sempre retornam True ou False

```python
print(10 > 3)    # True
print(10 < 3)    # False
print(10 >= 10)  # True
print(10 <= 9)   # False
print(10 == 10)  # True  → == compara; = atribui (não confunda!)
print(10 != 5)   # True  → diferente
```

## 3. Operadores Lógicos

Combinam condições

```python
tem_estoque = True
esta_disponivel = False

print(tem_estoque and esta_disponivel)  # False → ambos precisam ser True
print(tem_estoque or esta_disponivel)   # True  → basta um ser True
print(not tem_estoque)                  # False → inverte o valor
```

## 4. Operadores de Atribuição

```python
x = 10
x += 5   # x = x + 5  → 15
x -= 3   # x = x - 3  → 12
x *= 2   # x = x * 2  → 24
x //= 5  # x = x // 5 → 4
print(x)
```

# 📘 Aula 03 Strings

## NÍVEL 1 — Fundamentos | Aula 3: Strings

## 1. Criando strings

```python
simples = 'texto com aspas simples'
duplas = "texto com aspas duplas"
multilinhas = """
    Texto que
    ocupa várias linhas
"""
print(simples)
print(duplas)
print(multilinhas)
```

## 2. Operações básicas com strings

```python
nome = "Python"

print(len(nome))          # 6 → quantidade de caracteres
print(nome.upper())       # PYTHON
print(nome.lower())       # python
print(nome[0])            # P → primeiro caractere (índice começa em 0)
print(nome[-1])           # n → último caractere
print(nome[0:3])          # Pyt → fatiamento (slice): do índice 0 até 2

# Concatenação
saudacao = "Olá, " + nome + "!"
print(saudacao)

# Repetição
linha = "-" * 30
print(linha)
```

## 3. Métodos úteis de string

```python
frase = "  Bem-vindo ao treinamento Python!  "

print(frase.strip())             # remove espaços das bordas
print(frase.strip().upper())     # encadeando métodos
print(frase.replace("Python", "Data Engineering"))  # substitui texto
print(frase.strip().split(" "))  # divide em lista de palavras

email = "usuario@empresa.com"
print(email.startswith("usuario"))  # True
print(email.endswith(".com"))       # True
print("@" in email)                 # True → verifica se contém
```

## 4. Formatação de strings

```python
produto = "Notebook"
preco = 3599.99
quantidade = 3

# f-string (mais moderno e legível — use sempre este)
print(f"Produto: {produto} | Preço: R${preco:.2f} | Qtd: {quantidade}")

# Formatação de números
print(f"Valor total: R${preco * quantidade:,.2f}")  # 10,799.97
print(f"Percentual: {0.1574:.1%}")                  # 15.7%
```
