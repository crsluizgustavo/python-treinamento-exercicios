# 📘 Aula 01 Condicionais

## NÍVEL 2 — Controle de Fluxo | Aula 1: Condicionais

## 1. if / elif / else

Pense como a lógica que você já conhece:
SE condição ENTÃO faça X SENÃO faça Y

```python
temperatura = 35

if temperatura > 30:
    print("Está quente!")
elif temperatura >= 20:
    print("Temperatura agradável.")
else:
    print("Está frio!")
```

## 2. Condicionais aninhadas

```python
nota = 7.5
frequencia = 80

if frequencia >= 75:
    if nota >= 7:
        print("Aprovado")
    elif nota >= 5:
        print("Recuperação")
    else:
        print("Reprovado por nota")
else:
    print("Reprovado por falta")
```

## 3. Operadores lógicos em condicionais

Simplifica o exemplo acima

```python
if frequencia >= 75 and nota >= 7:
    print("Aprovado")
elif frequencia >= 75 and nota >= 5:
    print("Recuperação")
else:
    print("Reprovado")
```

## 4. Operador ternário (if em uma linha)

resultado = valor_se_verdadeiro if condição else valor_se_falso

```python
idade = 20
status = "Maior de idade" if idade >= 18 else "Menor de idade"
print(status)
```

## 5. Verificando tipo e conteúdo

```python
valor = None

if valor is None:
    print("Valor não informado")

lista = [1, 2, 3]
if lista:           # lista não vazia é True
    print("Lista tem itens")

texto = ""
if not texto:       # string vazia é False
    print("Texto vazio")
```

## 6. match / case (Python 3.10+) — equivalente ao switch

```python
dia = "segunda"

match dia:
    case "segunda" | "terça" | "quarta" | "quinta" | "sexta":
        print("Dia útil")
    case "sábado" | "domingo":
        print("Final de semana")
    case _:
        print("Dia inválido")
```

# 📘 Aula 02 Loops

## NÍVEL 2 — Controle de Fluxo | Aula 2: Loops

## 1. for — iterar sobre sequências

Use quando você sabe quantas vezes vai repetir

```python
frutas = ["maçã", "banana", "uva", "laranja"]

for fruta in frutas:
    print(fruta)

# range() → gera sequência de números
for i in range(5):          # 0, 1, 2, 3, 4
    print(i)

for i in range(1, 6):       # 1, 2, 3, 4, 5
    print(i)

for i in range(0, 10, 2):   # 0, 2, 4, 6, 8 (passo 2)
    print(i)

# enumerate() → índice + valor ao mesmo tempo
for indice, fruta in enumerate(frutas):
    print(f"{indice}: {fruta}")
```

## 2. while — repetir enquanto condição for verdadeira

Use quando você NÃO sabe quantas vezes vai repetir

```python
tentativas = 0
limite = 3

while tentativas < limite:
    print(f"Tentativa {tentativas + 1}")
    tentativas += 1

# Cuidado com loop infinito — sempre garanta que a condição vai se tornar False
```

## 3. break e continue

break → sai do loop imediatamente
continue → pula para a próxima iteração

```python
print("\n--- break ---")
for i in range(10):
    if i == 5:
        break
    print(i)   # imprime 0, 1, 2, 3, 4

print("\n--- continue ---")
for i in range(10):
    if i % 2 == 0:
        continue   # pula os pares
    print(i)       # imprime 1, 3, 5, 7, 9
```

## 4. for com else

O bloco else executa se o loop terminou SEM break

```python
numeros = [2, 4, 6, 8]
for n in numeros:
    if n % 2 != 0:
        print("Encontrou ímpar!")
        break
else:
    print("Todos os números são pares.")
```

## 5. Loops aninhados

```python
print("\n--- Tabuada do 1 ao 3 ---")
for i in range(1, 4):
    for j in range(1, 11):
        print(f"{i} x {j} = {i * j}")
    print()  # linha em branco entre tabuadas
```

## 6. Acumuladores — padrão muito comum

```python
vendas = [150.0, 230.5, 89.9, 410.0, 75.3]

total = 0
maior_venda = vendas[0]
menor_venda = vendas[0]

for venda in vendas:
    total += venda
    if venda > maior_venda:
        maior_venda = venda
    if venda < menor_venda:
        menor_venda = venda

media = total / len(vendas)
print(f"Total: R${total:.2f}")
print(f"Média: R${media:.2f}")
print(f"Maior venda: R${maior_venda:.2f}")
print(f"Menor venda: R${menor_venda:.2f}")
```

# 📘 Aula 03 Comprehensions

## NÍVEL 2 — Controle de Fluxo | Aula 3: Comprehensions

Comprehensions são uma forma concisa e pythônica de criar
listas, dicionários e conjuntos a partir de iterações.

## 1. List Comprehension

Sintaxe: [expressão for item in iterável if condição]

Jeito antigo (com loop)

```python
quadrados = []
for i in range(1, 6):
    quadrados.append(i ** 2)
print(quadrados)   # [1, 4, 9, 16, 25]

# Com list comprehension (equivalente — mais limpo)
quadrados = [i ** 2 for i in range(1, 6)]
print(quadrados)   # [1, 4, 9, 16, 25]

# Com filtro (if)
pares = [i for i in range(1, 11) if i % 2 == 0]
print(pares)   # [2, 4, 6, 8, 10]

# Transformando dados — muito usado em ETL!
precos_raw = ["R$29.90", "R$59.00", "R$14.50"]
precos = [float(p.replace("R$", "")) for p in precos_raw]
print(precos)   # [29.9, 59.0, 14.5]

# Normalizando nomes
nomes_raw = ["  JOÃO SILVA ", "  maria costa  ", "PEDRO ALVES "]
nomes = [n.strip().title() for n in nomes_raw]
print(nomes)   # ['João Silva', 'Maria Costa', 'Pedro Alves']
```

## 2. Dict Comprehension

Sintaxe: {chave: valor for item in iterável}

```python
produtos = ["arroz", "feijão", "milho"]
precos_lista = [5.90, 8.50, 3.20]

catalogo = {produto: preco for produto, preco in zip(produtos, precos_lista)}
print(catalogo)   # {'arroz': 5.9, 'feijão': 8.5, 'milho': 3.2}

# Invertendo um dicionário (chave ↔ valor)
invertido = {v: k for k, v in catalogo.items()}
print(invertido)

# Filtrando itens caros
caros = {k: v for k, v in catalogo.items() if v > 5}
print(caros)
```

## 3. Set Comprehension

```python
dados_duplicados = [1, 2, 2, 3, 3, 3, 4]
unicos = {x for x in dados_duplicados}
print(unicos)   # {1, 2, 3, 4}
```

## 4. Generator Expression

Como list comprehension, mas não carrega tudo na memória
Ideal para grandes volumes de dados

```python
soma = sum(i ** 2 for i in range(1, 1001))
print(f"Soma dos quadrados de 1 a 1000: {soma}")
```
