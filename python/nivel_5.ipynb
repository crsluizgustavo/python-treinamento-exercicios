# 📘 Aula 01 Classes E Objetos

## NÍVEL 5 — Orientação a Objetos | Aula 1: Classes e Objetos

Classe = molde / planta baixa
Objeto = instância criada a partir da classe

## 1. Definindo uma classe básica

```python
class Produto:
    # __init__ é o construtor — executado ao criar o objeto
    def __init__(self, nome, preco, estoque):
        self.nome = nome         # atributo de instância
        self.preco = preco
        self.estoque = estoque

    # Método de instância — opera sobre self (o próprio objeto)
    def valor_em_estoque(self):
        return self.preco * self.estoque

    def aplicar_desconto(self, percentual):
        self.preco = self.preco * (1 - percentual)

    # __str__ → define como o objeto aparece no print()
    def __str__(self):
        return f"Produto({self.nome}, R${self.preco:.2f}, estoque={self.estoque})"

    # __repr__ → representação técnica (usado no debug)
    def __repr__(self):
        return f"Produto(nome={self.nome!r}, preco={self.preco}, estoque={self.estoque})"


# Criando objetos (instâncias)
arroz = Produto("Arroz 5kg", 18.50, 100)
feijao = Produto("Feijão 1kg", 7.90, 250)

print(arroz)
print(feijao)
print(f"Valor em estoque do arroz: R${arroz.valor_em_estoque():.2f}")

arroz.aplicar_desconto(0.10)
print(f"Após desconto: {arroz}")

# Acessando e modificando atributos diretamente
feijao.estoque += 50
print(feijao)
```

## 2. Atributos de classe

```python
class Funcionario:
    empresa = "DataCorp"   # compartilhado por TODOS os funcionários
    total_contratados = 0

    def __init__(self, nome, cargo, salario):
        self.nome = nome
        self.cargo = cargo
        self.salario = salario
        Funcionario.total_contratados += 1

    def __str__(self):
        return f"{self.nome} | {self.cargo} | {self.empresa}"

f1 = Funcionario("Ana", "Data Engineer", 9500)
f2 = Funcionario("Bruno", "Data Analyst", 6200)
f3 = Funcionario("Carla", "Data Scientist", 8800)

print(f1)
print(f"Total contratados: {Funcionario.total_contratados}")
```

## 3. Métodos especiais (dunder methods)

```python
class Vetor:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vetor({self.x}, {self.y})"

    def __add__(self, outro):        # v1 + v2
        return Vetor(self.x + outro.x, self.y + outro.y)

    def __len__(self):               # len(v)
        return int((self.x**2 + self.y**2) ** 0.5)

    def __eq__(self, outro):         # v1 == v2
        return self.x == outro.x and self.y == outro.y

v1 = Vetor(2, 3)
v2 = Vetor(4, 1)
print(v1 + v2)     # Vetor(6, 4)
print(len(v1))     # 3
print(v1 == v2)    # False
```

# 📘 Aula 02 Heranca E Polimorfismo

## NÍVEL 5 — OOP | Aula 2: Herança e Polimorfismo

## 1. Herança básica

Herança = uma classe filha herda atributos e métodos da classe pai
Use quando existe relação "É UM" (Engenheiro É UM Funcionário)

```python
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def apresentar(self):
        return f"Sou {self.nome}, salário R${self.salario:,.0f}"

    def calcular_bonus(self):
        return self.salario * 0.05   # bônus padrão: 5%

    def __str__(self):
        return f"{self.__class__.__name__}: {self.nome}"


class Engenheiro(Funcionario):
    def __init__(self, nome, salario, linguagens):
        super().__init__(nome, salario)   # chama o __init__ do pai
        self.linguagens = linguagens

    def calcular_bonus(self):            # sobrescreve o método do pai (override)
        return self.salario * 0.15       # engenheiros ganham 15%

    def apresentar(self):
        base = super().apresentar()      # reutiliza o método do pai
        return f"{base} | Linguagens: {', '.join(self.linguagens)}"


class Gerente(Funcionario):
    def __init__(self, nome, salario, equipe):
        super().__init__(nome, salario)
        self.equipe = equipe    # lista de Funcionario

    def calcular_bonus(self):
        return self.salario * 0.20   # gerentes: 20%

    def total_equipe(self):
        return sum(f.salario for f in self.equipe)


# Criando objetos
ana = Engenheiro("Ana Lima", 9500, ["Python", "Spark", "SQL"])
bruno = Funcionario("Bruno Melo", 6200)
gerente = Gerente("Carla Souza", 12000, [ana, bruno])

print(ana.apresentar())
print(bruno.apresentar())
print(f"Bônus Ana: R${ana.calcular_bonus():.2f}")
print(f"Bônus Bruno: R${bruno.calcular_bonus():.2f}")
print(f"Bônus Carla: R${gerente.calcular_bonus():.2f}")
print(f"Total equipe: R${gerente.total_equipe():,.2f}")
```

## 2. Polimorfismo

Mesmo método, comportamentos diferentes dependendo do objeto
Permite tratar objetos diferentes de forma uniforme

```python
funcionarios = [
    Funcionario("Diego", 4500),
    Engenheiro("Elena", 11000, ["Python", "Kafka"]),
    Gerente("Fábio", 15000, []),
]

print("\n--- Folha de pagamento ---")
total_bonuses = 0
for f in funcionarios:
    bonus = f.calcular_bonus()   # cada classe calcula de forma diferente
    total_bonuses += bonus
    print(f"  {str(f):<30} Bônus: R${bonus:,.2f}")
print(f"  Total bônus: R${total_bonuses:,.2f}")
```

## 3. isinstance e issubclass

```python
print(isinstance(ana, Engenheiro))     # True
print(isinstance(ana, Funcionario))   # True — Engenheiro É UM Funcionario
print(isinstance(ana, Gerente))       # False
print(issubclass(Engenheiro, Funcionario))   # True
```

## 4. Classes abstratas (ABC)

```python
from abc import ABC, abstractmethod

class FonteDeDados(ABC):
    """Classe base abstrata — define o contrato para fontes de dados."""

    @abstractmethod
    def conectar(self):
        pass

    @abstractmethod
    def ler(self, query):
        pass

    def fechar(self):
        print(f"Conexão {self.__class__.__name__} encerrada.")


class PostgreSQL(FonteDeDados):
    def conectar(self):
        print("Conectando ao PostgreSQL...")

    def ler(self, query):
        print(f"Executando: {query}")
        return [{"id": 1, "nome": "Ana"}, {"id": 2, "nome": "Bruno"}]


class CSV(FonteDeDados):
    def __init__(self, caminho):
        self.caminho = caminho

    def conectar(self):
        print(f"Abrindo arquivo: {self.caminho}")

    def ler(self, query):
        print(f"Lendo CSV com filtro: {query}")
        return []


# FonteDeDados()  → TypeError: não pode instanciar classe abstrata

pg = PostgreSQL()
pg.conectar()
dados = pg.ler("SELECT * FROM usuarios")
print(dados)
pg.fechar()
```

# 📘 Aula 03 Encapsulamento E Propriedades

## NÍVEL 5 — OOP | Aula 3: Encapsulamento e Properties

Encapsulamento = controlar o acesso aos dados do objeto
Convenção Python: _atributo (protegido), __atributo (privado)

## 1. Atributos públicos, protegidos e privados

```python
class Sensor:
    def __init__(self, nome, valor_inicial):
        self.nome = nome             # público — acesso livre
        self._calibracao = 1.0       # protegido — use com cuidado externamente
        self.__leitura = valor_inicial   # privado — não acesse diretamente

    def get_leitura(self):
        return self.__leitura * self._calibracao

    def set_leitura(self, valor):
        if valor < 0:
            raise ValueError("Leitura não pode ser negativa")
        self.__leitura = valor

s = Sensor("Temperatura", 22.5)
print(s.get_leitura())
# print(s.__leitura)  → AttributeError (name mangling: _Sensor__leitura)
print(s._Sensor__leitura)   # tecnicamente possível, mas não faça isso
```

## 2. @property — a forma Pythônica de getters/setters

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self._preco = preco

    @property
    def preco(self):
        """Getter: chamado quando você acessa produto.preco"""
        return self._preco

    @preco.setter
    def preco(self, valor):
        """Setter: chamado quando você faz produto.preco = x"""
        if valor < 0:
            raise ValueError("Preço não pode ser negativo")
        self._preco = round(valor, 2)

    @property
    def preco_com_imposto(self):
        """Propriedade calculada — somente leitura"""
        return round(self._preco * 1.12, 2)

p = Produto("Notebook", 3500)
print(p.preco)              # 3500 — chama o getter
p.preco = 3299.99           # chama o setter
print(p.preco)              # 3299.99
print(p.preco_com_imposto)  # 3695.99

try:
    p.preco = -100
except ValueError as e:
    print(f"Erro: {e}")

# p.preco_com_imposto = 100   → AttributeError — sem setter definido
```

## 3. @classmethod e @staticmethod

```python
class Conexao:
    _instancias = 0
    _max_conexoes = 5

    def __init__(self, host, porta):
        if Conexao._instancias >= Conexao._max_conexoes:
            raise RuntimeError("Limite de conexões atingido")
        self.host = host
        self.porta = porta
        Conexao._instancias += 1

    def fechar(self):
        Conexao._instancias -= 1

    @classmethod
    def do_dsn(cls, dsn):
        """Cria conexão a partir de uma DSN string: host:porta"""
        host, porta = dsn.split(":")
        return cls(host, int(porta))

    @classmethod
    def conexoes_ativas(cls):
        return cls._instancias

    @staticmethod
    def validar_porta(porta):
        """Não precisa de self nem cls — é utilitário puro"""
        return 1 <= porta <= 65535

c1 = Conexao("localhost", 5432)
c2 = Conexao.do_dsn("db.empresa.com:5432")
print(f"Conexões ativas: {Conexao.conexoes_ativas()}")
print(f"Porta válida? {Conexao.validar_porta(5432)}")
print(f"Porta inválida? {Conexao.validar_porta(99999)}")
```
