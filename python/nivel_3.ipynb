# 📘 Aula 01 Listas E Tuplas

## NÍVEL 3 — Estruturas de Dados | Aula 1: Listas e Tuplas

## 1. Listas — mutáveis, ordenadas

Use quando precisar de uma coleção que pode mudar

```python
nomes = ["Ana", "Bruno", "Carlos"]
numeros = [1, 2, 3, 4, 5]
misto = [1, "texto", True, 3.14]    # pode misturar tipos (evite quando possível)
vazia = []

# Acesso por índice (começa em 0)
print(nomes[0])    # Ana
print(nomes[-1])   # Carlos (último)
print(nomes[1:3])  # ['Bruno', 'Carlos'] (slice: do índice 1 até 2)

# Modificando
nomes[1] = "Beatriz"
print(nomes)

# Métodos essenciais
nomes.append("Diego")           # adiciona no final
nomes.insert(1, "Artur")        # insere na posição 1
removido = nomes.pop()          # remove e retorna o último
nomes.remove("Artur")           # remove por valor (primeira ocorrência)
nomes.sort()                    # ordena in-place
nomes.reverse()                 # inverte in-place
copia = nomes.copy()            # cópia independente
print(f"Tamanho: {len(nomes)}")
print(f"Ana está na lista? {'Ana' in nomes}")
print(f"Índice de Ana: {nomes.index('Ana')}")
print(f"Ordenada: {sorted(nomes)}")   # retorna nova lista, não altera original

# Concatenação e repetição
lista_a = [1, 2, 3]
lista_b = [4, 5, 6]
combinada = lista_a + lista_b
print(combinada)          # [1, 2, 3, 4, 5, 6]

# Lista de listas (matriz)
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matriz[1][2])   # 6 → linha 1, coluna 2
```

## 2. Tuplas — imutáveis, ordenadas

Use quando os dados NÃO devem mudar: coordenadas, configurações, chaves compostas

```python
coordenadas = (23.5, -46.6)    # latitude, longitude de São Paulo
rgb_vermelho = (255, 0, 0)
versao = (3, 11, 4)

print(coordenadas[0])  # 23.5
# coordenadas[0] = 10  → TypeError! Tuplas são imutáveis

# Desempacotamento (muito útil!)
lat, lon = coordenadas
r, g, b = rgb_vermelho
print(f"Latitude: {lat}, Longitude: {lon}")

# Tupla com um único elemento (precisa da vírgula!)
unitaria = (42,)
print(type(unitaria))   # <class 'tuple'>

# Funções retornando múltiplos valores (internamente é uma tupla)
def minmax(lista):
    return min(lista), max(lista)

minimo, maximo = minmax([5, 2, 8, 1, 9])
print(f"Min: {minimo}, Max: {maximo}")
```

# 📘 Aula 02 Dicionarios

## NÍVEL 3 — Estruturas de Dados | Aula 2: Dicionários

Dicionário = estrutura chave:valor
É a estrutura mais usada em Data Engineering (JSON, APIs, configs)

## 1. Criando dicionários

```python
produto = {
    "nome": "Notebook Dell",
    "preco": 3599.99,
    "estoque": 12,
    "disponivel": True
}

# Acessando valores
print(produto["nome"])            # Notebook Dell
print(produto.get("preco"))       # 3599.99
print(produto.get("cor", "N/A"))  # N/A → valor padrão se chave não existir

# Modificando
produto["preco"] = 3299.99
produto["cor"] = "Preto"   # adiciona nova chave

# Removendo
del produto["disponivel"]
removido = produto.pop("cor")   # remove e retorna o valor
```

## 2. Iterando sobre dicionários

```python
print("\n--- Iterando ---")
for chave in produto:
    print(chave)

for chave, valor in produto.items():
    print(f"{chave}: {valor}")

print(list(produto.keys()))
print(list(produto.values()))
```

## 3. Verificações

```python
print("nome" in produto)       # True
print("cor" in produto)        # False
print(len(produto))
```

## 4. Dicionários aninhados

Muito comum para representar dados hierárquicos (JSON)

```python
funcionarios = {
    "F001": {
        "nome": "Ana Lima",
        "cargo": "Data Engineer",
        "salario": 8000,
        "habilidades": ["Python", "SQL", "Spark"]
    },
    "F002": {
        "nome": "Bruno Melo",
        "cargo": "Data Analyst",
        "salario": 6500,
        "habilidades": ["SQL", "Power BI", "Python"]
    }
}

# Acessando dados aninhados
print(funcionarios["F001"]["nome"])
print(funcionarios["F001"]["habilidades"][0])

# Iterando sobre estrutura aninhada
for codigo, dados in funcionarios.items():
    habilidades = ", ".join(dados["habilidades"])
    print(f"{codigo} | {dados['nome']} | {dados['cargo']} | R${dados['salario']:,} | {habilidades}")
```

## 5. Métodos úteis

```python
config_padrao = {"timeout": 30, "retries": 3, "verbose": False}
config_usuario = {"timeout": 60, "debug": True}

# update: mescla dicionários (o segundo sobrescreve o primeiro)
config_final = config_padrao.copy()
config_final.update(config_usuario)
print(config_final)

# setdefault: define valor apenas se chave não existir
config_final.setdefault("log_level", "INFO")
print(config_final)
```

# 📘 Aula 03 Sets

## NÍVEL 3 — Estruturas de Dados | Aula 3: Sets (Conjuntos)

Set = coleção SEM duplicatas, SEM ordem garantida
Use para: deduplicação, verificação de membros, operações de conjuntos

## 1. Criando sets

```python
tecnologias = {"Python", "SQL", "Spark", "Python", "SQL"}
print(tecnologias)   # {'Python', 'SQL', 'Spark'} — duplicatas removidas

vazio = set()   # NÃO use {} — isso cria dicionário vazio!

# Convertendo lista para set (remove duplicatas)
lista_com_dup = [1, 2, 2, 3, 3, 3, 4]
sem_dup = set(lista_com_dup)
print(sem_dup)
```

## 2. Operações de conjunto

```python
habilidades_ana   = {"Python", "SQL", "Spark", "Kafka"}
habilidades_bruno = {"SQL", "Power BI", "Python", "Tableau"}

# União: tudo que um ou outro tem
print(habilidades_ana | habilidades_bruno)
print(habilidades_ana.union(habilidades_bruno))

# Interseção: apenas o que ambos têm
print(habilidades_ana & habilidades_bruno)
print(habilidades_ana.intersection(habilidades_bruno))

# Diferença: o que Ana tem e Bruno não tem
print(habilidades_ana - habilidades_bruno)

# Diferença simétrica: o que só um deles tem
print(habilidades_ana ^ habilidades_bruno)

# Subconjunto?
basico = {"Python", "SQL"}
print(basico.issubset(habilidades_ana))   # True
print(habilidades_ana.issuperset(basico)) # True
```

## 3. Adicionando e removendo

```python
tecnologias.add("dbt")
tecnologias.discard("Spark")       # remove se existir, sem erro se não existir
print(tecnologias)
```

## 4. Caso de uso real: deduplicação de IDs

```python
ids_sistema_a = [101, 102, 103, 104, 102, 103]
ids_sistema_b = [103, 104, 105, 106, 107]

set_a = set(ids_sistema_a)
set_b = set(ids_sistema_b)

print(f"IDs em ambos os sistemas: {set_a & set_b}")
print(f"Apenas no sistema A: {set_a - set_b}")
print(f"Apenas no sistema B: {set_b - set_a}")
print(f"Total de IDs únicos: {len(set_a | set_b)}")
```
